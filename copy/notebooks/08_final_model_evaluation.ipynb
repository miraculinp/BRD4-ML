# 08 — Fresh training CV and locked-test evaluation

Stage 08 uses the Optuna best parameters for ExtraTrees and CatBoost, and scikit-learn default SVR parameters inside a `StandardScaler` pipeline. It first runs a fresh five-fold shuffled `GroupKFold` by scaffold on the training partition for all three models. Every fold independently drops incomplete, infinite, constant, and highly correlated features, then ranks/selects up to 250 features using that fold’s training rows. No imputation is used. These are the new CV results to report.

After CV, Stage 08 fits each final model on all training rows using the fixed 250-feature set from Stage 05, then evaluates once on the locked scaffold test set. CV uses a new fold allocation, but the hyperparameters were selected using the same training data in Stage 07, so the locked test remains the independent final generalization estimate. Do not retune or choose a model based on test scores.
